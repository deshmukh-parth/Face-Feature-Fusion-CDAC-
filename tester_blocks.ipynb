# testing for embedding formation using arcface

```python
import numpy as np
import cv2
import torch

from iresnet import iresnet50
```

```python
image = cv2.imread('person/img_1.jpg')
image_112 = cv2.resize(image, (112, 112))
```

```python
#convert to tensor [1, 3, 112, 112]
image_tensor = torch.from_numpy(image_112).permute(2, 0, 1).unsqueeze(0)
image_tensor = image_tensor.float().to("cpu")
```

```python
aface = iresnet50(fp16=False)
aface_weights =torch.load("arcface_w600k_r50.pth", map_location=torch.device('cpu'))
aface.load_state_dict(aface_weights)
aface =aface.to(device="cpu")
aface.eval()
```

```python
with torch.no_grad():
    embedding = aface(image_tensor)

embedding = embedding.cpu().numpy().flatten()

embedding = embedding / np.linalg.norm(embedding)

embedding.shape
```
